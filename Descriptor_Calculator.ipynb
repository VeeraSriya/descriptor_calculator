```python
from rdkit import Chem
from rdkit.Chem import Descriptors
```

```python
# Function to calculate descriptors
def calculate_descriptors(smiles):

    # Convert SMILES to molecule
    mol = Chem.MolFromSmiles(smiles)

    # Check invalid SMILES
    if mol is None:
        print("Invalid SMILES!")
        return

    # Calculate descriptors
    mw = Descriptors.MolWt(mol)
    logp = Descriptors.MolLogP(mol)
    tpsa = Descriptors.TPSA(mol)
    hbd = Descriptors.NumHDonors(mol)
    hba = Descriptors.NumHAcceptors(mol)

    # Print results
    print("\nDescriptor Results:\n")

    print("MW:", round(mw, 2))
    print("logP:", round(logp, 2))
    print("TPSA:", round(tpsa, 2))
    print("HBD:", hbd)
    print("HBA:", hba)


# Run program continuously
while True:

    smiles = input("\nEnter SMILES (or type quit): ")

    # Stop program
    if smiles.lower() == "quit":
        print("Program Ended")
        break

    # Calculate descriptors
    calculate_descriptors(smiles)
```
